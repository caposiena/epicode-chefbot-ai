# ChefBot AI - prova della ricerca semantica

In questa parte verifico la ricerca testuale sui 12 piatti presenti nella knowledge base.

La ricerca non controlla la presenza esatta delle parole. La richiesta e le descrizioni dei piatti vengono trasformate in embeddings e confrontate tramite cosine similarity.

```python
import sys
from pathlib import Path

PROJECT_ROOT = Path("..").resolve()
if str(PROJECT_ROOT) not in sys.path:
    sys.path.append(str(PROJECT_ROOT))

from src.knowledge_base import load_knowledge_base
from src.semantic_search import DishSearch
```

```python
dishes = load_knowledge_base("../data/knowledge_base.json")

print("Piatti caricati:", len(dishes))
print([dish["name"] for dish in dishes])
```

## Creazione degli embeddings

Uso un modello multilingua perché l'app riceve richieste in italiano, mentre alcune etichette originali di Food101 sono in inglese.

```python
search_engine = DishSearch(dishes)

print("Forma della matrice degli embeddings:", search_engine.embeddings.shape)
```

## Alcune richieste di prova

Per ogni frase visualizzo i tre risultati più vicini. In questo modo posso controllare se il primo suggerimento è sensato e osservare anche le alternative.

```python
queries = [
    "vorrei qualcosa di fresco e leggero",
    "ho voglia di un dolce cremoso al cucchiaio",
    "cerco un primo piatto sostanzioso con la pasta",
    "qualcosa da condividere in modo informale",
]

for query in queries:
    print(f"\nRichiesta: {query}")

    for result in search_engine.search(query, top_k=3):
        dish = result["dish"]
        score = result["score"]
        print(f"- {dish['name']}: {score:.3f}")
```

## Osservazione

La similarità indica quanto i testi sono vicini nello spazio degli embeddings, ma non rappresenta una probabilità. Con una knowledge base piccola il sistema restituisce comunque il piatto più vicino tra quelli disponibili, anche quando la richiesta è molto generica.

Per migliorare la ricerca si potrebbero ampliare le descrizioni, aggiungere più piatti e raccogliere alcune richieste reali con cui valutare i suggerimenti.
